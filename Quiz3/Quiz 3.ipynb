(b) Aggregate and query

<font size = "3">

- Import the "data_raw/sprint_results.csv" file
- Subset the observations for 'constructorId == 3' using ```.query()```
- Obtain the maximum of "laps" by "driverId" <br> 
 using ``` .groupby().agg() ```
- Subset the drivers with a maximum of laps $\ge 10$
- Store the results of all these operations in a dataframe <br>
```sprint_results_agg ```

````python
# Write your own code
sprint_results=pd.read_csv("data_raw/sprint_results.csv")

# aggregate
sprint_results.query("(constructorId==3)")
driveragg=(sprint_results.groupby('driverId').agg(
    maximum_laps=('laps','max')))
driveragg

# subset
sprint_results_agg=driveragg.query("(maximum_laps>=10)")
sprint_results_agg
````

(c) Recode a numeric colum

<font size = "3">

- Import the "data_raw/races.csv" file
- Recode the "year" column into a new column "year_brackets" <br>
with the following categories

$\qquad$ ``` ["1950-1965","1966-1980","1980-onwards"] ```

$\qquad$ HINT: Use the "pd.cut()" command.

````python
# Write your own code
races=pd.read_csv("data_raw/races.csv")
races['year']=pd.to_numeric(races['year'],errors='coerce') # ensure the year is numeric. 

# create bins & labels. 
bins_j=[1950,1965,1980,float('inf')]
labels_j=["1950-1965","1966-1980","1981-onward"]

# make categories. 
races['year_brackets']=pd.cut(races['year'],bins=bins_j,right=True,labels=labels_j)
races
````
